```python
dataset['diagnosis']=dataset['diagnosis'].map({'B': 0, 'M':1}).astype(int)
print(dataset)
```

```python
x=dataset.iloc[: , 2:32].values
x
```

```python
y=dataset.iloc[: , 1].values
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
print(x_train)
print(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(solver='liblinear',multi_class='ovr')  #loading algo
model.fit(x_train,y_train)  #train
value=[[20.57,17.77,132.9,1326,0.08474,0.07864,0.0869,0.07017,0.1812,0.05667,0.5435,0.7339,3.398,74.08,0.005225,0.01308,0.0186,0.0134,0.01389,0.003532,24.99,23.41,158.8,1956,0.1238,0.1866,0.2416,0.186,0.275,0.08902]]
y_pred=model.predict(value)
print(y_pred)

#The diagnosis of breast tissues (M = malignant: 1, B = benign : 0)

if y_pred==0:
  print("benign")
else:
  print("malignant")
```
